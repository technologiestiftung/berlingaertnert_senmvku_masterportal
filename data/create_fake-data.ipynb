```python
import json
import random
from shapely.geometry import Point, mapping

# Define an array of strings representing different garden types
garden_types = ["KitaSchulgaerten",
"Gartenarbeitsschule",
"Parklets",
"Beetpatenschaften",
"SolidarischeLandwirtschaft",
"Selbsterntegaerten" ,
"Flaechenpotentiale" ]

# Define Berlin bounding box (approximate latitude and longitude limits)
berlin_bbox = {
    "min_lon": 13.1601,
    "max_lon": 13.6126,
    "min_lat": 52.5819,
    "max_lat": 52.407
}

def generate_random_point():
    """Generate a random point within Berlin's bounding box."""
    lon = random.uniform(berlin_bbox["min_lon"], berlin_bbox["max_lon"])
    lat = random.uniform(berlin_bbox["min_lat"], berlin_bbox["max_lat"])
    return Point(lon, lat)

def create_geojson(garden_type):
    """Create a GeoJSON file with 100 random points for a given garden type."""
    features = []
    for _ in range(100):
        point = generate_random_point()
        feature = {
            "type": "Feature",
            "geometry": mapping(point),
            "properties": {"Gartentyp": garden_type}
        }
        features.append(feature)
    
    geojson = {
        "type": "FeatureCollection",
        "features": features
    }
    
    filename = f"../Berlingaertnert/resources/data/{garden_type.replace(' ', '_').lower()}.geojson"
    with open(filename, "w", encoding="utf-8") as f:
        json.dump(geojson, f, indent=4)
    print(f"GeoJSON file created: {filename}")

# Generate a GeoJSON file for each garden type
for garden_type in garden_types:
    create_geojson(garden_type)
```
